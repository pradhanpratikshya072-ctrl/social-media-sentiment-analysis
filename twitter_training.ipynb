```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("twitter_training.csv",header=None,names=["ID","Topic","Sentiment","Text"])
```

sentiment distribution 

```python
df["Sentiment"].value_counts()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,x="Sentiment",order=df["Sentiment"].value_counts().index)
plt.title("Overall Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Posts")
plt.show()
```

```python
df["Topic"].value_counts()
```

```python
top_topics=df["Topic"].value_counts().head(10).index
top_topics
```

```python
top_df=df[df["Topic"].isin(top_topics)]
top_df.head()
```

```python
plt.figure(figsize=(14,7))
sns.countplot(data=top_df,y="Topic",hue="Sentiment",order=top_topics)
plt.title("Sentiment Distribution Across Top 10 Topics/Brands")
plt.xlabel("Number of posts")
plt.ylabel("Topic / Brands")
plt.legend(title="Sentiment")
plt.show()
```

# Conclusion

The social media sentiment analysis was performed to understand public opinions and attitudes towards different topics and brands.

## Key Findings

- The dataset was cleaned by removing missing values and duplicate records before performing the analysis.
- The overall sentiment distribution shows that **Negative sentiment is the most common sentiment** in the dataset.
- **Positive sentiment** is the second most common, followed by **Neutral sentiment**.
- **Irrelevant sentiment** has the lowest number of posts among the four sentiment categories.
- The analysis of the top 10 topics/brands shows that sentiment patterns vary significantly across different topics.
- Some topics, such as **CallOfDuty**, show a particularly high number of negative posts, indicating stronger negative public reactions in the dataset.
- Other topics show a more balanced distribution of Positive, Neutral, Negative, and Irrelevant sentiments.
- Therefore, public opinion is not the same across all brands or topics, and sentiment analysis helps identify how audiences respond differently to each one.

## Final Conclusion

This analysis demonstrates that social media data can be effectively used to understand public opinion and attitudes towards different topics or brands. The results show an overall dominance of negative sentiment, while the topic-wise analysis reveals that audience reactions vary considerably across different topics. Such sentiment analysis can help organizations and brands understand customer perceptions and identify areas that may require attention or improvement.
